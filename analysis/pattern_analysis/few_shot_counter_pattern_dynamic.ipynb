```python
import os
import json
from statistics import mean

project_root = r"C:\Users\zakga\OneDrive\Documents\code\LeREaD_annotation_process"
folder = fr"{project_root}\data\Documents_Annotés\llm\F_PARACHUNKER_ALLINONE_p2_c500_fspattern-5-25_mgpt-5.2"

prefix = "selected_few_shot_examples"

doc_means = []
doc_details = {}

for filename in os.listdir(folder):
    if filename.startswith(prefix) and filename.endswith(".json"):
        filepath = os.path.join(folder, filename)

        with open(filepath, "r", encoding="utf-8") as f:
            data = json.load(f)

        # data = list of chunks
        # each chunk = list of examples (tuples)

        if not data:
            continue

        # number of examples per chunk
        examples_per_chunk = [len(chunk) for chunk in data if chunk]

        if not examples_per_chunk:
            continue

        doc_mean = mean(examples_per_chunk)
        doc_means.append(doc_mean)

        doc_details[filename] = {
            "num_chunks": len(data),
            "mean_examples_per_chunk": doc_mean
        }

# global mean across documents
global_mean = mean(doc_means) if doc_means else 0

# --- OUTPUT ---
print("Per-document means:")
for doc, stats in doc_details.items():
    print(f"{doc}: {stats}")

print("\nGlobal mean (across documents):", global_mean)
```
